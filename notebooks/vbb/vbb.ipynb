### Statische Daten (GTFS Downloads)
* Ein "Schnappschuss" des geplanten Angebots (Soll-Zustand).
* Alle Daten lokal, beliebig oft abfragen, ohne Rate-Limits der API zu fürchten, und sie eignen sich perfekt für langfristige Trendanalysen.

### Dynamische Echtzeit-Daten VBB-API (Rest-Schnittstelle)
* Live-Zugriff auf das System (Ist-Zustand + Auskunft).
* Echtzeit-Daten, die nicht in den statischen CSV-Dateien stehen.
* Endpunkte für dein Projekt:
    * departureBoard / arrivalBoard: Was fährt jetzt wirklich ab?
    * journeyPosition: Wo befindet sich ein Fahrzeug gerade exakt? (Perfekt für Live-Visualisierungen).
    * journeyDetail: Verspätungsinformationen und aktuelle Störungen.

### Dynamische Echtzeit-Daten VBB-GTFS-RT (Streaming-Feed)
* Kontinuierlicher Datenstrom: Ein binärer Feed, der den aktuellen "Herzschlag" des gesamten Netzes liefert (Ist-Zustand).
* Effizienz für Big Data: Im Gegensatz zur API-Abfrage einzelner Haltestellen liefert der Feed alle systemweiten Änderungen (Verspätungen, Fahrzeugpositionen) in einem kompakten Paket.
* Perfekt für Machine Learning: Da die Daten direkt über die trip_id mit dem statischen Fahrplan verknüpft sind, lassen sie sich ideal für Vorhersagemodelle und historische Verspätungsarchive nutzen.

* Kern-Komponenten für dein Projekt:
    * TripUpdates: Enthält die aktuellen Zeitabweichungen (Delays) für laufende Fahrten.
    * VehiclePositions: Liefert die GPS-Koordinaten der Fahrzeuge (ideal für Heatmaps der Netzbelastung).
    * ServiceAlerts: Informationen über außerplanmäßige Störungen, Baustellen oder Umleitungen.


### VBB GTFS Realtime Feed

The VBB GTFS Realtime feed can be found at the following URL:   
https://production.gtfsrt.vbb.de/data

You can inspect the feed in a web-based GTFS-RT inspector.   
It is licensed as CC-BY 4.0.
You can fetch it without authentication, up to a limit of 60 requests per minute.
Please make sure to subscribe to the Atom (RSS) feed to get notified about downtime, important changes, etc.
It has CORS enabled, so you can query it from any webpage.
It is served with an ETag header, allowing clients to cache it.
The code behind this is open.

## GTFS
Die General Transit Feed Specification (GTFS) ist ein offener Standard zum Austausch von Fahrplandaten des öffentlichen Verkehrs (ÖPNV) und geografischen Informationen (Haltestellen, Routen). Sie besteht aus Textdateien (.txt) und wird weltweit von Verkehrsbetrieben genutzt, um Daten für Apps wie Google Maps, Apple Maps oder den ÖPNV-Nutzer bereitzustellen. Es gibt zwei Hauptkomponenten: GTFS Static für Fahrpläne und GTFS Realtime für aktuelle Fahrdaten.

Documentations:

* Realtime: https://developers.google.com/transit/gtfs-realtime
* Static: https://developers.google.com/transit/gtfs/reference

* https://gtfs.org/getting-started/what-is-gtfs/
* https://gtfs.org/documentation/overview/


### GTFS Kern-Dateien (Stichwortliste)

| Datei | Kerninhalt | Funktion im Projekt |
| :--- | :--- | :--- |
| **agency.txt** | Verkehrsunternehmen | Identifikation (z.B. BVG vs. S-Bahn) via `agency_id`. |
| **routes.txt** | Linien-Stamm| Name (U7, M41) und Verkehrsmittel-Typ (Bus, Bahn). |
| **trips.txt** | Einzelfahrten | Verknüpft Routen mit Fahrplänen und Linienwegen (Shapes). |
| **stop_times.txt** | Ankunfts-/Abfahrtszeiten | Das Herzstück: Wann hält welcher Trip an welchem Stop? |
| **stops.txt** | Haltestellen | Geokoordinaten (Lat/Lon) und Namen der Stationen. |
| **shapes.txt** | Geometrie | Exakte Linienführung (Polylinien) für die Kartendarstellung. |
| **calendar.txt** | Wochentage | Gültigkeit: Fährt ein Trip Mo-Fr, Sa oder So? |
| **calendar_dates.txt**| Ausnahmen | Feiertage oder Sonderfahrten, die den Kalender überschreiben. |
| **transfers.txt** | Umstiegsinfos | Empfohlene Gehzeiten zwischen Bahnsteigen/Stationen. |
