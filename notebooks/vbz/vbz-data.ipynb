# Open Data - ÖPNV Stadt Zürich 
### Datenquellen, Strukturen und Strategie





## Überblick

Der ÖPNV in Zürich ist zweistufig organisiert:
- **ZVV** (Zürcher Verkehrsverbund) — Dachverband, koordiniert alle Linien
- **VBZ** (Verkehrsbetriebe Zürich) — operativer Betreiber, unser Hauptprotagonist

Für dieses Projekt werden vier Datentypen kombiniert:
Verkehrsdaten (Ist-Daten), Wetterdaten, Geodaten (GTFS) und Eventdaten.

### Archiv-URLs der Ist-Daten

Die Monatspakete folgen einem einheitlichen Schema:   
https://archive.opentransportdata.swiss/istdaten/{JAHR}/ist-daten-{JAHR}-{MM}.zip


**2023:** `ist-daten-2023-01.zip` bis `ist-daten-2023-12.zip`  
**2024:** `ist-daten-2024-01.zip` bis `ist-daten-2024-12.zip`  
**2025:** `ist-daten-2025-01.zip` bis `ist-daten-2025-12.zip`
**2026:** `ist-daten-2026-01.zip` bis `ist-daten-2026-XX.zip` *(laufend)*

> Hinweis: Ab Mitte 2025 wechselt das Format auf v2. Deshalb verwenden wir nur bis ende 2025 um einheitlich auf v1 zu arbeiten.

## Weitere Datenquellen

### Wetterdaten
- **Quelle:** [Stadt Zürich OGD](https://data.stadt-zuerich.ch/dataset/ugz_meteodaten_stundenmittelwerte)
- **Inhalt:** Stundenmittelwerte für Niederschlag, Temperatur, Wind
- **Stationen:** Fluntern (Hügellage) & Mythenquai (Seelage) — zwei Punkte für Topografie-Effekte
- **Format:** `.csv` → `meteo/`

### Geodaten (GTFS)
- **Quelle:** [ZVV / Stadt Zürich OGD](https://data.stadt-zuerich.ch/dataset/vbz_fahrplandaten_gtfs)
- **Inhalt:** Haltestellen-Koordinaten, Linienverläufe, Zuordnung zu Stadtkreisen
- **Format:** `.txt` (GTFS-Standard) → `gtfs/`
- **Notebook:** `vbz-gtfs.ipynb`

### Eventdaten
- **Quelle:** Manueller Crawl
- **Kategorien:** Stadion-Events, Stadtfeste, Konzerte, Messen, Schulferien
- **Format:** `.csv` → `events/`
- **Notebook:** `vbz-events.ipynb`




## Ausblick: Individualverkehr (v1.3)

Zürich betreibt ein Netz aus **Verkehrsinduktionsschleifen** (Zählstellen),
die kontinuierlich Fahrzeugdichte und Geschwindigkeit im Stadtgebiet messen.

- **Quelle:** [Stadt Zürich OGD – Verkehrszählung](https://data.stadt-zuerich.ch)
- **Mehrwert:** Verkehrsdichte (Autos/Std) als zusätzliches Feature im Modell
- **Ziel:** Zusammenhang zwischen Individualverkehr und Tram-Verspätungen sichtbar machen
